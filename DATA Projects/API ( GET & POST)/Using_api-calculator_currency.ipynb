```python
base_url = "https://v6.exchangerate-api.com/v6/c55a8ad40b814b1e3f78de50/latest/USD"
```

```python
import requests
```

```python
date = input("Please enter the date (yyyy-mm-dd or 'latest'): ")
base = input("Convert from (currency, e.g. USD): ").upper()
curr = input("Convert to (currency, e.g. EUR): ").upper()
quan = float(input(f"How much {base} do you want to convert: "))

# Build URL
url = f"{base_url}/{date}?base={base}&symbols={curr}"

# API request
response = requests.get(url)

if not response.ok:
    print(f"\n❌ Error {response.status_code}")
    print(response.json())
else:
    data = response.json()
    rate = data["rates"][curr]

    result = quan * rate

    print(f"{base} is equivalent to {curr} in {date}")
```
