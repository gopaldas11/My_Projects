```python
products = ["Product_A", "Product_B"]
sales = [223, 393, 202, 202, 922, 299]
time = (1,3,7)
for prod in products:
    for sale in sales:
        for tm in time:

            print(f"Expected sales for a period of {sale}:months for {prod}: ${sale*tm}")
```

```python
numbers = [23,44,66,76,44,33]
```

```python
lst_comp = [n * 2 for n in numbers]
lst_comp
```

```python
new_lst_compr = [i + j for i in range(2) for j in range(5)]
new_lst_compr
```

```python
new_lst_compp2 = [[i + j for i in range(2)] for j in range(5)]
new_lst_compp2
```

```python
# condition in the list comprehension loop
conditional_list_comp = [ num ** 3 for num in range(1,11) if num % 2 != 0]
conditional_list_comp
```

```python
test = [num ** 3 if num % 2 != 0 else "even" for num in range(1,11)]
test
```
