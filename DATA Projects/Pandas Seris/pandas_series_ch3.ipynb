**Pandas Series:.sort_values()**

```python
import pandas as pd
```

```python
numbers = pd.Series([98,44,65,78,24])
numbers
```

```python
numbers.sort_values()
```

```python
numbers.sort_values(ascending=False)
```

```python
numbers.sort_values(ascending=True)
```
